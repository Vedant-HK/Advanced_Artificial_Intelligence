```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate sample data
np.random.seed(32)
X = np.random.rand(20, 2)

# Create and fit GMM
n_components = 3
gmm = GaussianMixture(n_components=n_components)
gmm.fit(X)

# Predict clusters and probabilities
cluster_assignments = gmm.predict(X)
cluster_probabilities = gmm.predict_proba(X)

print("Cluster Assignments:")
print(cluster_assignments)

print("\nCluster Probabilities:")
print(cluster_probabilities)

# ---------------- PLOTTING ----------------

plt.figure(figsize=(7, 5))

# Scatter plot of data points colored by cluster
plt.scatter(
    X[:, 0], X[:, 1],
    c=cluster_assignments,
    s=40
)

# Plot cluster centers
means = gmm.means_
plt.scatter(
    means[:, 0], means[:, 1],
    s=200,
    marker='X'
)

plt.title("Gaussian Mixture Model Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
