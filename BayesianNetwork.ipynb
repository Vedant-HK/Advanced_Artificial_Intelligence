```python
!pip install pgmpy

# Import necessary libraries
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Create Bayesian Network
model = DiscreteBayesianNetwork()

# Define the variables
# D = Disease, S1 = Symptom1, S2 = Symptom2, T = Test
model.add_nodes_from(['D', 'S1', 'S2', 'T'])

# Define the structure of the network
model.add_edge('D', 'S1')
model.add_edge('D', 'S2')
model.add_edge('D', 'T')

# Define the CPDs

cpd_d = TabularCPD(
    variable='D',
    variable_card=2,
    values=[[0.7],
            [0.3]]
)

cpd_s1 = TabularCPD(
    variable='S1',
    variable_card=2,
    values=[[0.8, 0.1],
            [0.2, 0.9]],
    evidence=['D'],
    evidence_card=[2]
)

cpd_s2 = TabularCPD(
    variable='S2',
    variable_card=2,
    values=[[0.6, 0.2],
            [0.4, 0.8]],
    evidence=['D'],
    evidence_card=[2]
)

cpd_t = TabularCPD(
    variable='T',
    variable_card=2,
    values=[[0.9, 0.3],
            [0.1, 0.7]],
    evidence=['D'],
    evidence_card=[2]
)

# Add CPDs to the model
model.add_cpds(cpd_d, cpd_s1, cpd_s2, cpd_t)

# Validate the model
assert model.check_model()

# Perform inference
inference = VariableElimination(model)

# Example evidence
evidence = {'S1': 1, 'S2': 0, 'T': 1}

posterior_prob = inference.query(
    variables=['D'],
    evidence=evidence
)

print("Posterior Probability of Disease:")
print(posterior_prob)
```
